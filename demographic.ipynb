```python
nb_n_cols = [col for col in df.columns if col.startswith('NB_') and col != 'NBPERS_HOUSE' and col != 'NB_CAR']
check = df['NBPERS_HOUSE'] == df[nb_n_cols].sum(axis=1)
all(check)

print(df['NBPERS_HOUSE'])
print(df[nb_n_cols].sum(axis=1))
```

```python
cols_to_print = ['NBPERS_HOUSE'] + nb_n_cols
print(df[cols_to_print])
```

```python
df['DRIVING_LICENCE'].dropna().unique()
```

```python
priority_dict = {
    'No diploma': 0,
    "5-year-and-above higher education degree: Master's 2, DEA, DESS, Grande École Diploma, Doctorate, etc.": 5,
    '3–4-year higher education degree: Licence, Professional Licence, Master 1, or equivalent': 4,
    'Lower secondary certificate (Brevet) or equivalent': 1,
    'Upper secondary diploma (Baccalauréat) or equivalent': 3,
    'Vocational certificate (CAP, BEP) or equivalent': 2
}

df['DIPLOMA'] = df['DIPLOMA'].map(priority_dict)
df.drop(columns=['TYPE_HOUSE', 'AREA_NAME'], inplace=True)

df = df.applymap(lambda x: float(x) if isinstance(x, (int, float)) else (1.0 if x is True else (0.0 if x is False else x)))
df['SEX'] = df['SEX'].map({'Man': 1.0, 'Woman': 0.0})
df['DIPLOMA'] = df['DIPLOMA'].fillna(0)
df.head()
```

```python
df.to_csv("data/individuals_dataset_cleaned.csv", index=False)
```
