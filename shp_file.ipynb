```python
import geopandas as gpd

arr = gpd.read_file("arrondissements/arrondissements.shp").to_crs(epsg=4326)

# 1) Número de arrondissements únicos
n = arr["l_ar"].nunique()
print(f"Hay {n} arrondissements distintos en el shapefile.")

# 2) Listado de sus nombres
print("Nombres de arrondissement:")
print(arr["l_ar"].unique())
```

```python
import os

total_lines = 0
folder = "data/gps_filtrados"

for filename in os.listdir(folder):
    if filename.endswith(".csv"):
        with open(os.path.join(folder, filename), 'r', encoding='utf-8') as f:
            total_lines += sum(1 for _ in f)

print(f"Total number of lines in all CSV files: {total_lines}")
```
